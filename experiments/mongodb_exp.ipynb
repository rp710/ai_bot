```python
import os
from dotenv import load_dotenv, find_dotenv
from pymongo import MongoClient

# Load .env file from project root
load_dotenv(find_dotenv())
```

```python
uri = os.getenv("MONGODB_URI")
```

```python
client = MongoClient(uri)

# Verify connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. Connected successfully to MongoDB!")
except Exception as e:
    print(f"Connection error: {e}")
```

```python
db = client["test_ai_bot"]
collection = db["products"]
print("connected to mongodb")
```

```python
product = {
    "name":"shirt",
    "price":99,
    "category":"men"
}
collection.insert_one(product)
```

```python
products = [
    {"name": "Jeans", "price": 1499, "category": "men"},
    {"name": "Kurti", "price": 999, "category": "women"},
    {"name": "Kids Jacket", "price": 1299, "category": "kids"}
]

collection.insert_many(products)

print("Multiple Products Inserted ✅")
```

```python
all_products = collection.find()

for p in all_products:
    print(p)
```

```python
product = collection.find_one({"name": "Kurti"})

print(product)
```

```python
collection.update_one(
    {"name": "Jeans"},
    {"$set": {"price": 1399}}
)

print("Product Updated ✅")
```

```python
collection.delete_one({"name": "Kids Jacket"})

print("Product Deleted ✅")
```

```python
collection.delete_many({})

print("All Products Deleted ✅")
```

```python
collection.drop()
print("collection dropped")
```
